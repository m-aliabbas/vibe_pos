```python
import pandas as pd
```

```python
df = pd.read_csv("customers.csv")
```

```python
# df remove the rows where MobileNo is less than 10 digits
df = df[df['MobileNo'].astype(str).str.len() >= 10]
```

```python
# fill with empty string where NaN
df = df.fillna('')
```

```python
df
```

```python
df_deb = df[df['AccountType'] == 'DEBITORS']
```

```python
df_cred = df[df['AccountType'] == 'CREDITORS']
```

```python
df_cred.to_csv("customers_creditors.csv", index=False)
```

```python
df_deb.to_csv("vendor_debitors.csv", index=False)
```
